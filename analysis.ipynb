```python
top_platforms = df.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=False).head(10)
```

```python
top_platforms
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_platforms.index, top_platforms.values)

plt.title("En Çok Satış Yapan 10 Oyun Platformu")
plt.xlabel("Platform")
plt.ylabel("Global Sales (Milyon)")

plt.grid(axis="y", alpha=0.3)
plt.savefig("top_platforms.png")

plt.show()
```

```python
top_genres = df.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_genres.index, top_genres.values)

plt.title("En Çok Satış Yapan 10 Oyun Türü")
plt.xlabel("Genre")
plt.ylabel("Global Sales (Milyon)") 

plt.grid(axis="y", alpha=0.3)
plt.savefig("top_genres.png")

plt.show()
```

```python
year_sales = year_df.groupby("Year")["Global_Sales"].sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,5))
plt.bar(year_sales.index, year_sales.values)
plt.title("Yıllara Göre Global Satışlar")
plt.xlabel("Year")
plt.ylabel("Global Sales (Milyon)")

plt.grid(axis="y", alpha=0.3)
plt.savefig("year_sales.png")

plt.show()
```

```python
regional_sales = df[["NA_Sales","EU_Sales","JP_Sales","Other_Sales"]].sum()
```

```python
plt.figure(figsize=(8,5))
plt.bar(regional_sales.index, regional_sales.values)

plt.title("Bölgelere Göre Toplam Oyun Satışları")
plt.xlabel("Bölge")
plt.ylabel("Global Sales (Milyon)")

plt.grid(axis="y", alpha=0.3)
plt.savefig("regional_sales.png")

plt.show()
```
